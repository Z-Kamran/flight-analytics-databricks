```python
%sql
SELECT
    f.id AS ID,

    -- Airports
    TRIM(f.departure_airport) AS Dept_Airport,
    TRIM(f.arrival_airport) AS Arr_Airport,

    -- Dates
    TO_DATE(from_utc_timestamp(f.scheduled_time, 'America/Chicago')) AS Sch_Dept_Date,
    TO_DATE(from_utc_timestamp(f.actual_time, 'America/Chicago')) AS Actual_Dept_Date,

      -- Time Only
    date_format(
        from_utc_timestamp(f.scheduled_time, 'America/Chicago'),
        'HH:mm'
    ) AS Sch_Dept_Time,
    date_format(
        from_utc_timestamp(f.actual_time, 'America/Chicago'),
        'HH:mm'
    ) AS Actual_Dept_Time,

    -- Timestamps (Central Time)
    from_utc_timestamp(f.scheduled_time, 'America/Chicago') AS Sch_Dept_Time_CT,
    from_utc_timestamp(f.actual_time, 'America/Chicago') AS Actual_Dept_Time_CT,

    -- Flight Details
    CONCAT(f.airlines_iata_code, f.flight_number) AS Flight_Number,
    TRIM(f.flight_status) AS Flight_Status,
    TRIM(f.gate) AS Gate,
    TRIM(f.aircraft_type) AS Aircraft,
    
    -- Airline Lookup
   a.Airline

FROM groupbwt_airport_flight_delay_schedule_change_dataset.flights.flights f

LEFT JOIN data_from_kaggle.bronze.airlinecodes a
    ON f.airlines_iata_code = a.IATA;
```

```python
%sql
CREATE TABLE IF NOT EXISTS flights_schema.silver.flights_list_table_silver
AS
SELECT
    f.id AS ID,

    -- Airports
    TRIM(f.departure_airport) AS Dept_Airport,
    TRIM(f.arrival_airport) AS Arr_Airport,

    -- Dates
    TO_DATE(from_utc_timestamp(f.scheduled_time, 'America/Chicago')) AS Sch_Dept_Date,
    TO_DATE(from_utc_timestamp(f.actual_time, 'America/Chicago')) AS Actual_Dept_Date,

      -- Time Only
    date_format(
        from_utc_timestamp(f.scheduled_time, 'America/Chicago'),
        'HH:mm'
    ) AS Sch_Dept_Time,
    date_format(
        from_utc_timestamp(f.actual_time, 'America/Chicago'),
        'HH:mm'
    ) AS Actual_Dept_Time,

    -- Timestamps (Central Time)
    from_utc_timestamp(f.scheduled_time, 'America/Chicago') AS Sch_Dept_Time_CT,
    from_utc_timestamp(f.actual_time, 'America/Chicago') AS Actual_Dept_Time_CT,

    -- Flight Details
    CONCAT(f.airlines_iata_code, f.flight_number) AS Flight_Number,
    TRIM(f.flight_status) AS Flight_Status,
    TRIM(f.gate) AS Gate,
    TRIM(f.aircraft_type) AS Aircraft,
    
    -- Airline Lookup
   a.Airline

FROM groupbwt_airport_flight_delay_schedule_change_dataset.flights.flights f

LEFT JOIN data_from_kaggle.bronze.airlinecodes a
    ON f.airlines_iata_code = a.IATA;
```

```python
%sql
--- percentage of delayed routes
WITH route_totals AS (
    SELECT
        Dept_Airport,
        Arr_Airport,
        COUNT(*) AS Total_Flights
    FROM flights_schema.silver.flights_list_table_silver
    WHERE Sch_Dept_Time_CT IS NOT NULL
      AND Actual_Dept_Time_CT IS NOT NULL
    GROUP BY Dept_Airport, Arr_Airport
),

route_delays AS (
    SELECT
        Dept_Airport,
        Arr_Airport,
        COUNT(*) AS Delayed_Flights
    FROM flights_schema.silver.flights_list_table_silver
    WHERE Sch_Dept_Time_CT IS NOT NULL
      AND Actual_Dept_Time_CT IS NOT NULL
      AND (
            unix_timestamp(Actual_Dept_Time_CT)
            - unix_timestamp(Sch_Dept_Time_CT)
          ) / 60 > 0
    GROUP BY Dept_Airport, Arr_Airport
)

SELECT
    CONCAT(t.Dept_Airport, ' - ', t.Arr_Airport) AS Route,
    t.Total_Flights,
    COALESCE(d.Delayed_Flights, 0) AS Delayed_Flights,
    ROUND(
        COALESCE(d.Delayed_Flights, 0) * 100.0 /
        t.Total_Flights,
        2
    ) AS Delay_Percentage
FROM route_totals t
LEFT JOIN route_delays d
    ON t.Dept_Airport = d.Dept_Airport
   AND t.Arr_Airport = d.Arr_Airport
ORDER BY Delay_Percentage DESC;
```

```python
%sql
--- Best Performance score, where every delayed flight is -0.5 and every on-time flight is +1
WITH airline_stats AS (
    SELECT
        Airline,
        COUNT(*) AS Total_Flights,
        SUM(
            CASE
                WHEN (unix_timestamp(Actual_Dept_Time_CT) -
                      unix_timestamp(Sch_Dept_Time_CT)) / 60 > 0
                THEN 1
                ELSE 0
            END
        ) AS Delayed_Flights
    FROM flights_schema.silver.flights_list_table_silver
    WHERE Sch_Dept_Time_CT IS NOT NULL
      AND Actual_Dept_Time_CT IS NOT NULL AND Airline IS NOT NULL
    GROUP BY Airline
)

SELECT
    Airline,
    Total_Flights,
    Delayed_Flights,
    ((Total_Flights)*1 - (Delayed_Flights)*0.5) AS Performance_Score
FROM airline_stats
ORDER BY Performance_Score DESC
LIMIT 5;
```
